```python
import urllib
import requests
from pathlib import Path as path
from bs4 import BeautifulSoup
```

```python
url:str = 'https://openngo.ru/opendata/'
```

```python
#def obtem_dados_russia()->path:
    
    #realiza o download dos dados da organicao sem fim lucrativo da Russia

    #:paramentos url:endereco do site com dados da organicao sem fim lucrativo da Russia
    #return: caminho para dados de saida

html = urllib.request.urlopen(url).read()
soup = BeautifulSoup(html, 'html.parser')
```

```python
tags = soup.find_all('a', href=True)

link = []
for a in tags[11:77-7]:
    link.append(a['href'])
```

```python
caminho_saida = path('../dados/ONGS')
caminho_saida.mkdir(parents=True, exist_ok=True)
```

```python
for dado in link:
    r = requests.get(link[11], stream=True)
    with open(caminho_saida/"ONGS.zip", 'wb') as arq:
        arq.write(r.content)
```
